```python
import os
```

```python
data_directory = '/users/sgdbareh/volatile/ECHR_Importance/Art_3_Data/corpora/article3'
```

```python
list_files = []

for folder in os.listdir(data_directory):

    subfolder = data_directory + '/' + folder

    for file in os.listdir(subfolder):
        list_files.append(subfolder+'/'+file)
```

```python
facts_list = []
law_list = []

for file in list_files:
    
    if file.split('/')[-1] == 'fact_section':
        for text_file in os.listdir(file):
            with open(file + '/' + text_file,'r') as text:
                data = [text_file, text.read()]
            facts_list.append(data)

    else:
        for text_file in os.listdir(file):
            with open(file + '/' + text_file,'r') as text:
                data = [text_file, text.read()]
            law_list.append(data)
```

```python
facts_df = pd.DataFrame(facts_list,columns=['Filename','Facts'])
law_df = pd.DataFrame(law_list,columns=['Filename','The Law'])
```

```python
facts_df['Filename'] = facts_df['Filename'].str.replace('.txt', '')
law_df['Filename'] = law_df['Filename'].str.replace('.txt', '')
facts_df['Facts'] = facts_df['Facts'].str.strip()
law_df['The Law'] = law_df['The Law'].str.strip()
facts_df['Facts'] = facts_df['Facts'].str.replace('\n', '')
law_df['The Law'] = law_df['The Law'].str.replace('\n', '')
```

```python
#create lists from sclappnos and extractedappno
full_outcome_data['sclappnos_list'] = full_outcome_data['sclappnos'].apply(lambda x: x.split(';') if pd.notnull(x) else [])
full_outcome_data['extractedappno_list'] = full_outcome_data['extractedappno'].apply(lambda x: x.split(';') if pd.notnull(x) else [])

full_outcome_data['differences'] = full_outcome_data.apply(lambda row: list(set(row['sclappnos_list']) - set(row['extractedappno_list'])), axis=1)

full_outcome_data[['sclappnos', 'extractedappno', 'differences','appno']].head(50)
```

```python
#this cleans the list of differences to remove any empty strings or NaNs and returns the length of the list of the list

def clean_list(x):
    if isinstance(x, list):
        cleaned = [item for item in x if pd.notna(item) and item.strip()]
        return len(cleaned)
    else:
        return 0

full_outcome_data['diff_len'] = full_outcome_data['differences'].apply(clean_list)
```

```python
full_outcome_data.reset_index(drop=True, inplace=True)
```

```python
#adds in the missing sclappnos to the extractedappno list
full_outcome_data['extractedappno_list'] = full_outcome_data.apply(lambda row: row['extractedappno_list'] + row['differences'] if row['diff_len'] > 0 else row['extractedappno_list'], axis=1)
```

```python
#removes the appno of the case itself from the cited appnos
full_outcome_data['extractedappno_list'] = full_outcome_data.apply(lambda row: [x for x in row['extractedappno_list'] if x not in row['appno']], axis=1)
```

```python
full_outcome_data['extractedapplen'] = full_outcome_data['extractedappno_list'].apply(lambda x: len(x))
```

```python
full_outcome_data[['extractedappno_list','appno','extractedapplen','differences']].head(50)
```

```python
full_outcome_data['extractedapplen'].plot(kind='hist', bins=50)

plt.xlabel('Length of extractedappno')
plt.ylabel('Frequency')
plt.title('Distribution of extractedappno Length')

plt.show()
```

```python
full_outcome_data.drop(['diff_len','differences','sclappnos_list'],axis=1,inplace=True)
```

```python
#create list of appnos assigned to each case
full_outcome_data['appno_list'] = full_outcome_data['appno'].apply(lambda x: x.split(';'))
```

```python
#assign first appno to case
full_outcome_data['first_app_no'] = full_outcome_data['appno_list'].apply(lambda x: x[0])
```

```python
# helps to standardize the appnos  - if the appno is in appno_list, replace it with the first appno

for index, row in full_outcome_data.iterrows():
    extractedappno_list = row['extractedappno_list']
    #print(extractedappno_list)
    print(index)
    for i in range(len(extractedappno_list)):
        print('old ', extractedappno_list[i])
        try:
            full_outcome_row = full_outcome_data[full_outcome_data['appno_list'].apply(lambda x: extractedappno_list[i] in x)].iloc[0]
            extractedappno_list[i] = full_outcome_row['first_app_no']
            print('new ', extractedappno_list[i])
        except:
            pass    

    full_outcome_data.iloc[index]['extractedappno_list'] = extractedappno_list
            
```

```python
#remove app no dupes
full_outcome_data['extactedapppno_dupe'] = full_outcome_data['extractedappno_list'].apply(lambda x: list(set(x)))
```

```python
#give len of no dupe list app nos
full_outcome_data['new_list_length'] = full_outcome_data['extactedapppno_dupe'].apply(lambda x: len(x))
```
